**Mål:** Hent data fra nettsider med requests + BeautifulSoup.

## Recap fra forelesning

**Web scraping** = automatisk henting av data fra nettsider.

### Prosessen
1. **Send forespørsel** — `requests.get(url)` henter HTML
2. **Parse HTML** — `BeautifulSoup(html, 'html.parser')` lager et parse-tre
3. **Naviger treet** — `soup.find()`, `soup.find_all()`, CSS-selektorer
4. **Trekk ut data** — `.text`, `.get('href')`, `.get('src')`

### Viktige BeautifulSoup-metoder
| Metode | Hva den gjør |
|--------|-------------|
| `soup.find('tag')` | Finner **første** element |
| `soup.find_all('tag')` | Finner **alle** elementer |
| `soup.select('css')` | CSS-selektor (`.class`, `#id`) |
| `element.text` | Henter ren tekst |
| `element.get('attr')` | Henter attributtverdi |

### Etikk
- Sjekk `robots.txt` før du scraper
- Legg til pause mellom forespørsler (`time.sleep()`)
- Identifiser deg med en User-Agent header
