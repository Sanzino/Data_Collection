## Recap fra forelesning

**ETL = Extract → Transform → Load**

| Steg | Hva skjer | pandas-funksjoner |
|------|----------|-------------------|
| **Extract** | Hent data fra kilder (CSV, JSON, Excel) | `read_csv()`, `read_json()`, `read_excel()` |
| **Transform** | Rens, filtrer, berik, kombiner | `dropna()`, `drop_duplicates()`, `merge()`, `apply()` |
| **Load** | Lagre resultat | `to_csv()`, `to_parquet()` |

**Bronze → Silver → Gold:**
- **Bronze**: Rå data, uendret fra kilde
- **Silver**: Renset, validert, standardisert
- **Gold**: Aggregert, klar for analyse/rapportering
