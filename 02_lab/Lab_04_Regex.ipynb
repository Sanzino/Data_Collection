## Recap fra forelesning

### Viktige funksjoner
| Funksjon | Hva den gjør |
|----------|-------------|
| `re.match(p, s)` | Matcher fra **starten** av strengen |
| `re.search(p, s)` | Finner **første** match |
| `re.findall(p, s)` | Returnerer **alle** matcher som liste |
| `re.sub(p, r, s)` | **Erstatter** alle matcher |

### Vanlige mønstre
| Mønster | Betyr |
|---------|-------|
| `\d` | Siffer (0-9) |
| `\w` | Bokstav/siffer/underscore |
| `\s` | Whitespace |
| `.` | Hva som helst (unntatt newline) |
| `+` | 1 eller flere |
| `*` | 0 eller flere |
| `{n}` | Nøyaktig n ganger |
| `[abc]` | a, b, eller c |
| `(...)` | Capture group |

## Oppgave 2: E-postvalidering

Skriv et regex-mønster som matcher gyldige e-postadresser. Test mot disse:

```python
tester = [
    "anna@test.no",       # Gyldig
    "erik.hansen@firma.com",  # Gyldig
    "@mangler.no",        # Ugyldig
    "test@.com",          # Ugyldig
    "a+b@test.co.uk",     # Gyldig
]
```

> **Hint:** Et enkelt e-post-mønster: `[\w.+-]+@[\w-]+\.[a-zA-Z]{2,}`

````python
tester = [
    "anna@test.no",
    "erik.hansen@firma.com",
    "@mangler.no",
    "test@.com",
    "a+b@test.co.uk",
]

# Ditt mønster her
epost_pattern = r''

for t in tester:
    match = re.match(epost_pattern, t)
    status = 'GYLDIG' if match else 'UGYLDIG'
    print(f"  {t:30} -> {status}")
````
