## Recap fra forelesning

**API** = Application Programming Interface — en strukturert måte å hente data fra tjenester.

### HTTP-metoder
| Metode | Handling | Eksempel |
|--------|---------|----------|
| `GET` | Hent data | Hent brukerinfo |
| `POST` | Send/opprett data | Lag ny bruker |
| `PUT` | Oppdater (hele) | Erstatt brukerprofil |
| `DELETE` | Slett | Fjern bruker |

### Statuskoder
| Kode | Betydning |
|------|-----------|
| `200` | OK |
| `201` | Created |
| `400` | Bad Request |
| `401` | Unauthorized (mangler nøkkel) |
| `404` | Not Found |
| `429` | Too Many Requests (rate limit) |
| `500` | Server Error |

### Vanlig mønster
```python
import requests
response = requests.get('https://api.example.com/data')
if response.status_code == 200:
    data = response.json()  # Parse JSON til dict/list
```
