## 1. Hva er data?

**Data** er fakta, observasjoner eller målinger som kan brukes til analyse.

Data finnes overalt — fra sensorer og sosiale medier til helseregistre og finanstransaksjoner.

> "The basic idea behind the phrase 'Big Data' is that everything we do is increasingly leaving a digital trace (or data), which we can use and analyze to become smarter." — Bernard Marr

### Datatyper: Kvalitativ vs. Kvantitativ

| Type | Beskrivelse | Eksempler |
|------|------------|----------|
| **Kvalitativ (kategorisk)** | Beskriver egenskaper, ikke tall | Farge, kjønn, mening |
| **Kvantitativ (numerisk)** | Målbare verdier | Alder, temperatur, inntekt |

**Kvalitativ** deles videre i:
- **Nominal**: Ingen naturlig rekkefølge (f.eks. farge, land)
- **Ordinal**: Har rekkefølge (f.eks. lav/middels/høy)

**Kvantitativ** deles videre i:
- **Diskret**: Heltall, tellbare (f.eks. antall barn)
- **Kontinuerlig**: Kan ha alle verdier (f.eks. temperatur, høyde)

```python
import pandas as pd
import json
import numpy as np
import os
import datetime

# Eksempel: Klassifisere datatyper
data = {
    'Navn': ['Anna', 'Erik', 'Sara', 'Ole'],           # Kvalitativ - Nominal
    'Utdanning': ['Bachelor', 'Master', 'PhD', 'VGS'],  # Kvalitativ - Ordinal
    'Alder': [25, 32, 28, 45],                           # Kvantitativ - Diskret
    'Hoyde_cm': [165.2, 180.5, 172.0, 175.8]             # Kvantitativ - Kontinuerlig
}
df = pd.DataFrame(data)
display(df)

print("\nDatatyper i DataFrame:")
print(df.dtypes)
```

### Måleskalaer (NOIR)

De fire måleskalaene (fra lavest til høyest presisjon):

| Skala | Beskrivelse | Eksempel | Operasjoner |
|-------|------------|----------|-------------|
| **Nominal** | Kun kategorier, ingen orden | Blodtype (A, B, AB, O) | ==, != |
| **Ordinal** | Kategorier med orden | Tilfredshet (1-5 stjerner) | ==, !=, <, > |
| **Intervall** | Lik avstand, ingen ekte nullpunkt | Temperatur i Celsius | +, - |
| **Ratio** | Lik avstand + ekte nullpunkt | Vekt, høyde, inntekt | +, -, *, / |

```python
# Demonstrere måleskalaer med konkrete data
maleskalaer = {
    'Nominal': {'data': ['A', 'B', 'AB', 'O', 'A', 'B'], 'beskrivelse': 'Blodtype'},
    'Ordinal': {'data': [1, 3, 5, 2, 4, 3], 'beskrivelse': 'Tilfredshet (1-5)'},
    'Intervall': {'data': [20.5, -5.0, 37.2, 0.0, 15.3, -10.0], 'beskrivelse': 'Temperatur (C)'},
    'Ratio': {'data': [75.0, 82.5, 60.0, 95.2, 70.1, 88.0], 'beskrivelse': 'Vekt (kg)'}
}

for skala, info in maleskalaer.items():
    print(f"{skala}: {info['beskrivelse']} -> {info['data']}")

# Ratio: vi KAN si at 90 kg er dobbelt så tungt som 45 kg
# Intervall: vi kan IKKE si at 20°C er dobbelt så varmt som 10°C
print("\n--- Ratio vs Intervall ---")
print(f"90 kg / 45 kg = {90/45} (meningsfullt)")
print(f"20°C / 10°C = {20/10} (IKKE meningsfullt — 0°C er ikke 'ingen varme')")
```

## 2. Strukturert vs. Ustrukturert vs. Semi-strukturert data

| Type | Beskrivelse | Format | Eksempler |
|------|------------|--------|----------|
| **Strukturert** | Organisert i rader/kolonner | Tabeller, SQL-databaser | CSV, Excel, relasjonsdatabaser |
| **Semi-strukturert** | Har noe struktur, men fleksibelt | Tags/nøkler | JSON, XML, HTML, log-filer |
| **Ustrukturert** | Ingen fast struktur | Fri-form | Bilder, videoer, e-poster, PDF-er |

> Ca. 80% av all data er ustrukturert!

```python
# STRUKTURERT: CSV/tabell
print("=== STRUKTURERT (tabell) ===")
strukturert = pd.DataFrame({
    'id': [1, 2, 3],
    'navn': ['Anna', 'Erik', 'Sara'],
    'alder': [25, 32, 28]
})
display(strukturert)

# SEMI-STRUKTURERT: JSON
print("\n=== SEMI-STRUKTURERT (JSON) ===")
semi_strukturert = {
    "ansatt": {
        "navn": "Anna",
        "alder": 25,
        "ferdigheter": ["Python", "SQL", "Pandas"],
        "adresse": {"by": "Oslo", "postnr": "0150"}
    }
}
print(json.dumps(semi_strukturert, indent=2))

# USTRUKTURERT: fri tekst
print("\n=== USTRUKTURERT (fri tekst) ===")
ustrukturert = """Hei! Jeg lurer på om det er mulig å bestille 
3 stk laptops til IT-avdelingen. Budsjettet er 
ca. 45000 kr totalt. Mvh, Ole"""
print(ustrukturert)
print("\nUtfordring: Hvordan trekke ut 'antall=3', 'produkt=laptop', 'budsjett=45000'?")
```

## 3. Analog vs. Digital data

- **Analog data**: Kontinuerlige signaler (lydbølger, temperatur i naturen)
- **Digital data**: Diskrete verdier representert som 0-er og 1-ere

All data må digitaliseres (ADC - Analog to Digital Conversion) for å kunne prosesseres av datamaskiner.

```python
# Simuler analog-til-digital konvertering
timer = np.linspace(0, 24, 1000)  # 1000 punkter (analog-aktig)
temp_analog = 15 + 8 * np.sin((timer - 6) * np.pi / 12)

# Digital: bare måler hvert 3. time (8 målinger)
timer_digital = np.arange(0, 24, 3)
temp_digital = 15 + 8 * np.sin((timer_digital - 6) * np.pi / 12)

print("Analog signal: 1000 målinger (kontinuerlig)")
print(f"Digital sampling: {len(timer_digital)} målinger (diskret)")
print(f"\nDigitale temperaturmålinger:")
for t, temp in zip(timer_digital, temp_digital):
    print(f"  Kl {t:02.0f}:00 -> {temp:.1f}°C")
```

## 4. Datastørrelser

| Enhet | Verdi | Eksempel |
|-------|-------|----------|
| Byte (B) | 8 bits | Et enkelt tegn |
| Kilobyte (KB) | 1 000 B | Et kort tekstdokument |
| Megabyte (MB) | 1 000 KB | Et bilde, en MP3-sang |
| Gigabyte (GB) | 1 000 MB | En HD-film |
| Terabyte (TB) | 1 000 GB | Stor harddisk |
| Petabyte (PB) | 1 000 TB | Facebook genererer ~4 PB/dag |
| Exabyte (EB) | 1 000 PB | All data på internett (~463 EB/dag innen 2025) |
| Zettabyte (ZB) | 1 000 EB | Global datalagring ~44 ZB i 2020 |

```python
# Beregn datastørrelser
storrelse_bytes = {
    'Byte': 1,
    'Kilobyte': 1_000,
    'Megabyte': 1_000_000,
    'Gigabyte': 1_000_000_000,
    'Terabyte': 1_000_000_000_000,
    'Petabyte': 1_000_000_000_000_000,
}

print("Hvor mange MP3-sanger (5 MB) får plass på ulike lagringsmedier?")
mp3_size = 5_000_000  # 5 MB
for enhet, bytes_val in storrelse_bytes.items():
    if bytes_val >= mp3_size:
        antall = bytes_val // mp3_size
        print(f"  1 {enhet}: {antall:>15,} sanger")

# Sjekk størrelsen på vår egen datafil
data_path = '../04_data/sales_data.csv'
if os.path.exists(data_path):
    size = os.path.getsize(data_path)
    print(f"\nVåre egne filer:")
    print(f"  sales_data.csv: {size:,} bytes ({size/1000:.1f} KB)")
```

## 5. Big Data

**Big Data** = datasett som er for store eller komplekse for tradisjonelle verktøy.

### Small Data vs. Big Data

| | Small Data | Big Data |
|---|-----------|----------|
| **Størrelse** | KB — GB | TB — ZB |
| **Verktøy** | Excel, SQL | Hadoop, Spark |
| **Struktur** | Strukturert | Alle typer |
| **Hastighet** | Sakte akkumulert | Kontinuerlig strøm |
| **Eksempel** | Regneark med kundedata | Social media feeds, IoT-sensorer |

### De 4 V-ene (og 7 V-er)

| V | Definisjon | Eksempel |
|---|-----------|----------|
| **Volume** | Enorme datamengder | Facebook: petabytes daglig |
| **Variety** | Mange ulike formater | Tekst, bilder, video, sensorer |
| **Velocity** | Høy hastighet på datainnsamling | Twitter: 500M tweets/dag |
| **Veracity** | Datakvalitet og pålitelighet | Falske nyheter, sensor-feil |
| **Variability** | Data endrer seg over tid | Sesongvariasjoner i e-handel |
| **Visualization** | Presentasjon av innsikt | Dashboards i Tableau/Power BI |
| **Value** | Forretningsverdi fra data | Anbefalingssystemer (Amazon, Netflix) |

```python
# Demonstrer Volume-konseptet med et reelt datasett
df_auto = pd.read_csv('../04_data/auto_data.csv')
print(f"auto_data.csv: {df_auto.shape[0]} rader x {df_auto.shape[1]} kolonner")
print(f"Minnebruk: {df_auto.memory_usage(deep=True).sum() / 1024:.1f} KB")

print("\n--- Tenk deg Big Data-skala ---")
skala = [1, 1_000, 1_000_000, 1_000_000_000]
for s in skala:
    minne = (df_auto.memory_usage(deep=True).sum() / 1024) * s / df_auto.shape[0]
    enhet = 'KB' if minne < 1000 else 'MB' if minne < 1_000_000 else 'GB'
    verdi = minne if enhet == 'KB' else minne/1000 if enhet == 'MB' else minne/1_000_000
    print(f"  {s:>13,} rader -> ~{verdi:,.1f} {enhet}")
```

## 6. Modern Data Ecosystem

Et moderne data-økosystem består av:

1. **Data Sources** — Sosiale medier, sensorer, databaser, IoT
2. **Data Management** — Innsamling, lagring, organisering
3. **Users** — Business stakeholders, analytikere, utviklere, data scientists

Rå data må **organiseres, renses, optimaliseres** og følge standarder for organisasjonen.

### Data Pipeline

En data pipeline flytter data fra ett sted/format til et annet.

**Steg i en pipeline:**
1. **Data Extraction** — Hent fra kilder (DB, API, filer)
2. **Data Ingestion** — Overføring til pipeline (batch eller real-time)
3. **Transformation** — Rensing, filtrering, berikelse
4. **Loading** — Lagre i måldatabase/warehouse
5. **Scheduling** — Automatisere kjøring
6. **Monitoring** — Overvåke ytelse og kvalitet
7. **Maintenance** — Vedlikehold og optimalisering

```python
# Simuler en enkel data pipeline
log = []

def logg(steg, melding):
    tidspunkt = datetime.datetime.now().strftime("%H:%M:%S")
    log.append(f"[{tidspunkt}] {steg}: {melding}")
    print(log[-1])

# 1. EXTRACT
logg("EXTRACT", "Leser sales_data.csv")
df = pd.read_csv('../04_data/sales_data.csv')
logg("EXTRACT", f"Hentet {len(df)} rader")

# 2. TRANSFORM
logg("TRANSFORM", "Sjekker manglende verdier")
missing = df.isnull().sum().sum()
logg("TRANSFORM", f"Fant {missing} manglende verdier")

logg("TRANSFORM", "Fjerner duplikater")
before = len(df)
df = df.drop_duplicates()
logg("TRANSFORM", f"Fjernet {before - len(df)} duplikater")

# 3. LOAD
output_path = '../06_output/pipeline_demo.csv'
df.to_csv(output_path, index=False)
logg("LOAD", f"Lagret {len(df)} rader til {output_path}")

print("\n--- Pipeline fullført! ---")
```

### Batch vs. Stream Pipeline

| | Batch Pipeline | Stream Pipeline |
|---|---------------|----------------|
| **Prosessering** | I bolker (timer/dager) | Kontinuerlig, real-time |
| **Latens** | Høy (minutter-timer) | Lav (millisekunder-sekunder) |
| **Bruksområde** | Backup, rapporter, analyse | Fraud detection, live dashboards |
| **Verktøy** | Cron jobs, Airflow | Kafka, Spark Streaming |
| **Eksempel** | Nattlig lønnkjøring | Kredittkort-svindel-deteksjon |

### Pipeline Performance

- **Throughput**: Hvor mye data kan prosesseres per tidsenhet (records/sec)
- **Latency**: Tid fra data sendes til den er tilgjengelig i mål-systemet

> Lavere latens = raskere innsikt, men potensielt lavere datakvalitet

### Lambda Architecture
Kombinerer **batch** og **streaming** i ett system:
- **Batch Layer**: Prosesserer historisk data (Hadoop)
- **Speed Layer**: Real-time prosessering (Kafka, Flink)
- **Serving Layer**: Kombinerer begge for spørringer (NoSQL, Elasticsearch)

## 7. Data Storage

### Database
Organisert samling av data, styrt av et DBMS (Database Management System).
- Bruker SQL for spørringer
- Populære: PostgreSQL, MySQL, Oracle, SQL Server

### OLTP vs. OLAP

| | OLTP | OLAP |
|---|------|------|
| **Formål** | Transaksjoner | Analyse |
| **Spørringer** | Enkle, raske | Komplekse, sakte |
| **Data** | Normalisert, mange tabeller | Denormalisert, færre tabeller |
| **Eksempel** | "Hvem kjøpte X?" | "Hvor mange kjøpte X?" |

### Data Warehouse vs. Data Lake

| Parameter | Data Lake | Data Warehouse |
|-----------|-----------|----------------|
| **Datatype** | Rå, alle typer | Prosessert, strukturert |
| **Formål** | Ukjent / TBD | Kjent, i bruk |
| **Prosess** | ELT | ETL |
| **Brukere** | Data scientists | Business-analytikere |
| **Størrelse** | Petabytes | Terabytes |

## 8. Filformater

| Format | Type | Struktur | Best for |
|--------|------|----------|----------|
| **CSV** | Rad-basert | Ren tekst | Enkle datasett, ETL |
| **JSON** | Semi-strukturert | Tekst | API-er, web-data |
| **XML** | Semi-strukturert | Tekst | Konfigurasjon, legacy |
| **Parquet** | Kolonne-basert | Binær | Analytics, big data |
| **ORC** | Kolonne-basert | Binær | Hadoop/Hive |
| **Avro** | Rad-basert | Binær | Streaming (Kafka) |

```python
# Demonstrer ulike filformater i Python

# CSV
print("=== CSV ===")
df_small = pd.DataFrame({
    'Manufacturer': ['Acura', 'BMW', 'Toyota'],
    'Model': ['Integra', '328i', 'Camry'],
    'Price': [21500, 33400, 24000]
})
df_small.to_csv('/tmp/demo.csv', index=False)
print(open('/tmp/demo.csv').read())

# JSON
print("=== JSON ===")
json_data = df_small.to_dict(orient='records')
print(json.dumps(json_data, indent=2))

# XML
print("\n=== XML ===")
xml = '<?xml version="1.0"?>\n<cars>\n'
for _, row in df_small.iterrows():
    xml += f'  <car>\n    <manufacturer>{row.Manufacturer}</manufacturer>\n'
    xml += f'    <model>{row.Model}</model>\n'
    xml += f'    <price>{row.Price}</price>\n  </car>\n'
xml += '</cars>'
print(xml)

# Sammenlign filstørrelser
print("\n=== Filstørrelse-sammenligning ===")
df_small.to_csv('/tmp/demo.csv', index=False)
df_small.to_json('/tmp/demo.json', orient='records')
df_small.to_parquet('/tmp/demo.parquet', index=False)
for fmt in ['csv', 'json', 'parquet']:
    size = os.path.getsize(f'/tmp/demo.{fmt}')
    print(f"  {fmt.upper():>8}: {size:>6} bytes")
```

### Row Storage vs. Columnar Storage

| Aspekt | Row Storage | Column Storage |
|--------|------------|----------------|
| **Lagring** | Alle felt i en rad sammen | Alle verdier i en kolonne sammen |
| **Best for** | OLTP (inserts, updates) | OLAP (aggregeringer) |
| **Systemer** | MySQL, PostgreSQL | Parquet, Amazon Redshift |
| **Komprimering** | Lavere | Høyere (samme datatype) |

## 9. Datakilder

### Tre hovedkategorier:

1. **Relasjonsdatabaser** — Interne systemer (SQL Server, Oracle, MySQL)
2. **Flat-filer og XML** — Offentlige og private datasett (CSV, Excel, XML)
3. **API-er og Web Services** — Programmatisk tilgang (REST APIs, Web Scraping)
4. **Data Streams** — Real-time feeds (Kafka, Spark, IoT-sensorer)

### API vs. Web Scraping

| Feature | API | Web Scraping |
|---------|-----|--------------|
| Tilgang | Offisiell, strukturert | Uoffisiell, ustrukturert |
| Format | JSON, XML | HTML (krever parsing) |
| Pålitelighet | Høy | Lav (avhenger av nettside) |
| Legalitet | Generelt lovlig | Kan bryte vilkår |
